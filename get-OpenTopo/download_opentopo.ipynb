This notebook is used to download topography data from OpenTopography (https://opentopography.org/)
It can be used to download an individual tiff or to download multiple tiffs 

## Imports

```python
from GetOpenTopo import GetOpenTopo # import from the GetOpenTopo.py in same folder
```

## Definitions

Define file location, bounds of individual GeoTiffs, type of data and maximum extend with spacing if downloading multiple geotiffs along longitude

```python
bounds = [-37.05, -16.85, 112.45, 114.45]                                      # south, north, west, east of tile
output_directory = '../Data/Copernicus_OpenTopo_Australia/'                    # where to store GeoTIFF(s)
demtype = 'COP30'                                                              # which dataset to use
api_key = 'demoapikeyot2022'                                                   # OpenTopo API KEY (use yours)

dlon = 2                                                                       # longitude width of GeoTIFF (for seamless tiles bounds should be dlon apart)
max_lon = 152.45                                                               # define a maximum longitude to be the last upper boundary to download
geotiff_num = round((max_lon - (bounds[-1])+2)/dlon)                           # number of geotiffs to be downloaded if multiple
```

## Dowload GeoTiffs

Version to just download one tiff

```python
gt = GetOpenTopo(bounds, output_directory, demtype, api_key)
gt.download_topography()
```

Version to download multiple based on dlon defined above

```python
bounds_iter = bounds.copy()
counter = 0
while not (bounds_iter[-1] > max_lon):
    gt = GetOpenTopo(bounds_iter, output_directory, demtype, api_key)
    gt.download_topography()
    bounds_iter[-2] += dlon
    bounds_iter[-1] += dlon
    counter += 1

    print('Downloaded {} from {}. Last file downloaded: {}'.format(counter+1, str(round(max_lon - (bounds[-1]))+2), gt.get_filename()), end='\r')
```
